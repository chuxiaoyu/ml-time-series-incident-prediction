## Check the dataset and select a subset that contains enough anomalies.

```python
import pandas as pd
from pathlib import Path

base_dir = Path("data/CINECA/raw_datasets")

for csv_path in sorted(base_dir.glob("*/dataset_*.csv")):
    df = pd.read_csv(csv_path)
    if "label" not in df.columns:
        print(f"{csv_path}: no 'label' column")
        continue
    counts = df["label"].value_counts().sort_index()
    total = len(df)
    anomaly_count = counts.get(1, 0)
    anomaly_pct = anomaly_count / total * 100 if total > 0 else 0.0

    print(f"=== {csv_path} ===")
    print(counts)
    print(f"Anomaly percentage (label==1): {anomaly_pct:.2f}%\n")
```

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

base_dir = Path("data/CINECA")

rows = []
for split in ["train", "val", "test"]:
    csv_path = base_dir / f"{split}.csv"
    df = pd.read_csv(csv_path)
    counts = df["y"].value_counts().sort_index()
    total = len(df)
    anom = counts.get(1, 0)
    rows.append(
        {
            "split": split,
            "total": total,
            "nom": counts.get(0, 0),
            "anom": anom,
            "anom_pct": f"{(anom / total * 100) if total > 0 else 0.0:05.2f}%",
        }
    )

label_dist_df = pd.DataFrame(rows)
display(label_dist_df)

# Also show as markdown for easy copy-paste
from IPython.display import Markdown
md = label_dist_df.to_markdown(index=False)
display(Markdown(md))
```

## check evaluation results

```python
from pathlib import Path
import pandas as pd
from IPython.display import Markdown, display

results_dir = Path("results")

all_results = []
for csv_path in sorted(results_dir.glob("*.csv")):
    df = pd.read_csv(csv_path)
    # Add filename to identify the model/result source
    df.insert(0, "file", csv_path.name)
    all_results.append(df)

if all_results:
    results_df = pd.concat(all_results, ignore_index=True)
    display(results_df)

    # Also show as markdown table for easy copy-paste into README, etc.
    results_df = results_df[["model", "precision", "recall", "f1", "train_time_sec"]]
    for col in ["precision", "recall", "f1", "train_time_sec"]:
        results_df[col] = results_df[col].round(4)
    md = results_df.to_markdown(index=False)
    display(Markdown(md))
else:
    print("No results found in 'results/' directory.")
```
